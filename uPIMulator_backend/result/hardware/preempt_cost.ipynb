```python
import pandas as pd

def create_initial_df():

    df = pd.DataFrame(columns=['code'])
    print("创建了初始DataFrame，包含code列")
    return df
def add_column_to_df(df, column_name, code, default_value):
    
    df = df.copy()  # 避免修改原DataFrame
    
    # 检查列是否存在
    if column_name in df.columns:
        mask = df['code'] == code
        if mask.any():
            df.loc[mask, column_name] = default_value
        else:
            print(f"未找到code='{code}'的行")
    else:
        # 添加新列，默认值为空或NaN
        df[column_name] = pd.NA
        # 为特定code设置值
        mask = df['code'] == code
        if mask.any():
            df.loc[mask, column_name] = default_value
        else:
            print(f"未找到code='{code}'的行")
    
    return df
```

```python
import re
def extract_time_from_line(line):
    pattern = r'preempt.*?finish.*?with\s+(\d+)'
    
    match = re.search(pattern, line, re.IGNORECASE)  # 忽略大小写
    if match:
        return int(match.group(1))  # 返回第一个捕获组（数字部分）
    return None
def extract_time_from_txt_file(filename,df,code):
    counter=0
    try:
        with open(filename, 'r', encoding='utf-8', errors='ignore') as file:
            for line_num, line in enumerate(file, 1):
                line = line.strip()
                time_value = extract_time_from_line(line)
                
                if time_value:
                    df=add_column_to_df(df, "preempt_"+str(counter), code, time_value);
                    counter=counter+1;
                
            
    except Exception as e:
        print(f"读取文件 {filename} 时出错: {e}")
    return df

```

```python
import os
import glob
from pathlib import Path
import csv
import pandas as pd
import re
#
"""使用os.listdir遍历当前目录中包含'txt'的文件"""
df=create_initial_df()
for filename in os.listdir("25"):
    if 'txt' in filename:
        name_part = filename.split('_')[0]
        new_row = pd.DataFrame({'code': [str(name_part)]})
        df = pd.concat([df, new_row], ignore_index=True)
        df=extract_time_from_txt_file("25/"+filename,df,str(name_part))
for filename in os.listdir("10"):
    if 'txt' in filename:
        name_part = filename.split('_')[0]
        new_row = pd.DataFrame({'code': [str(name_part)]})
        df = pd.concat([df, new_row], ignore_index=True)
        df=extract_time_from_txt_file("10/"+filename,df,str(name_part))
for filename in os.listdir("5"):
    if 'txt' in filename:
        name_part = filename.split('_')[0]
        new_row = pd.DataFrame({'code': [str(name_part)]})
        df = pd.concat([df, new_row], ignore_index=True)
        df=extract_time_from_txt_file("5/"+filename,df,str(name_part))      
df.to_csv('preempt.csv', index=False, encoding='utf-8')
```
