```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import re
import os
from collections import Counter

# Set font for matplotlib
plt.rcParams['font.sans-serif'] = ['DejaVu Sans', 'Arial']
plt.rcParams['axes.unicode_minus'] = False

def read_txt_and_extract_start_times(txt_file_path):
    """
    Read txt file and extract all DMA start timestamps, sorted by time
    
    Args:
    txt_file_path: path to txt file
    
    Returns:
    list: sorted list of DMA start timestamps
    """
    try:
        with open(txt_file_path, 'r', encoding='utf-8') as file:
            lines = file.readlines()
        
        print(f"Successfully read txt file: {txt_file_path}")
        print(f"Total lines: {len(lines)}")
        
        start_times = []
        
        # Regex pattern to match "start dma" followed by numbers
        pattern = r'start dma (\d+)'
        
        print("Extracting DMA start timestamps...")
        
        for line_num, line in enumerate(lines, 1):
            matches = re.findall(pattern, line, re.IGNORECASE)
            
            for match in matches:
                timestamp = int(match)
                start_times.append(timestamp)
                if len(start_times) <= 10:  # Show only first 10
                    print(f"   Line {line_num}: found start dma {timestamp}")
        
        if len(start_times) > 10:
            print(f"   ... (showing first 10, found {len(start_times)} total)")
        
        # Sort by timestamp
        start_times.sort()
        print(f"Extraction complete, found {len(start_times)} DMA start timestamps")
        print(f"Time range: {start_times[0]} - {start_times[-1]}")
        
        return start_times
        
    except Exception as e:
        print(f"Error reading txt file: {e}")
        return []

def calculate_intervals(start_times):
    """
    Calculate time interval between each start_dma and the previous start_dma
    
    Args:
    start_times: sorted list of DMA start timestamps
    
    Returns:
    list: list of time intervals
    """
    if len(start_times) < 2:
        print("Need at least 2 timestamps to calculate intervals")
        return []
    
    intervals = []
    
    print("Calculating time intervals...")
    print("First 10 interval details:")
    
    for i in range(1, len(start_times)):
        interval = start_times[i] - start_times[i-1]
        intervals.append(interval)
        
        if i <= 10:  # Show first 10 intervals
            print(f"   Interval {i}: {start_times[i]} - {start_times[i-1]} = {interval}")
    
    if len(intervals) > 10:
        print(f"   ... (showing first 10, calculated {len(intervals)} intervals)")
    
    return intervals

def analyze_intervals_by_buckets(intervals, bucket_size=10):
    """
    Analyze interval distribution by specified bucket size
    
    Args:
    intervals: list of time intervals
    bucket_size: size of each bucket, default 10
    
    Returns:
    dict: dictionary containing statistical analysis results
    """
    if not intervals:
        return {}
    
    total_count = len(intervals)
    
    # Calculate max interval to determine number of buckets needed
    max_interval = max(intervals)
    num_buckets = (max_interval // bucket_size) + 1
    
    print(f"Interval statistical analysis (grouped by {bucket_size} units)")
    print(f"Total intervals: {total_count}")
    print(f"Max interval: {max_interval}")
    print(f"Number of buckets: {num_buckets}")
    
    # Initialize statistical results
    stats = {
        'bucket_size': bucket_size,
        'total_count': total_count,
        'max_interval': max_interval,
        'buckets': {},
        'cumulative': {},
        'percentages': {},
        'cumulative_percentages': {}
    }
    
    # Calculate statistics for each bucket
    print(f"Detailed distribution statistics:")
    print(f"{'Range':<15} {'Count':<8} {'Ratio':<8} {'Cumulative':<10} {'Cum Ratio'}")
    print("-" * 55)
    
    cumulative_count = 0
    
    for i in range(num_buckets):
        lower_bound = i * bucket_size
        upper_bound = (i + 1) * bucket_size
        
        # Calculate number of intervals in current bucket range
        if i == num_buckets - 1:  # Last bucket includes values equal to upper_bound
            count = sum(1 for interval in intervals if lower_bound < interval <= upper_bound)
            range_str = f"({lower_bound}, {upper_bound}]"
        else:
            count = sum(1 for interval in intervals if lower_bound < interval <= upper_bound)
            range_str = f"({lower_bound}, {upper_bound}]"
        
        if i == 0:  # First bucket includes 0
            count = sum(1 for interval in intervals if 0 <= interval <= upper_bound)
            range_str = f"[0, {upper_bound}]"
        
        if count > 0:  # Only record non-empty buckets
            percentage = (count / total_count) * 100
            cumulative_count += count
            cumulative_percentage = (cumulative_count / total_count) * 100
            
            stats['buckets'][range_str] = count
            stats['cumulative'][range_str] = cumulative_count
            stats['percentages'][range_str] = percentage
            stats['cumulative_percentages'][range_str] = cumulative_percentage
            
            print(f"{range_str:<15} {count:<8} {percentage:<7.2f}% {cumulative_count:<10} {cumulative_percentage:<7.2f}%")
    
    return stats

def find_matching_files(folder_a, folder_b):
    """
    Find matching txt files in two folders
    
    Args:
    folder_a: path to first folder
    folder_b: path to second folder
    
    Returns:
    list: list of matching filenames (without path)
    """
    if not os.path.exists(folder_a) or not os.path.exists(folder_b):
        print("Error: One or both folders do not exist")
        return []
    
    files_a = set([f for f in os.listdir(folder_a) if f.endswith('.txt')])
    files_b = set([f for f in os.listdir(folder_b) if f.endswith('.txt')])
    
    matching_files = files_a.intersection(files_b)
    
    print(f"Found {len(matching_files)} matching txt files:")
    for file in sorted(matching_files):
        print(f"   {file}")
    
    return list(matching_files)

def create_comparison_visualization(folder_a, folder_b, filename, output_image_name, bucket_size=10, max_intervals=100):
    """
    Create comparison visualization for cumulative distribution from two folders
    
    Args:
    folder_a: path to first folder
    folder_b: path to second folder
    filename: name of the txt file to compare
    output_image_name: name for saved image
    bucket_size: size of each bucket, default 10
    max_intervals: maximum number of intervals to display, default 100
    """
    file_a_path = os.path.join(folder_a, filename)
    file_b_path = os.path.join(folder_b, filename)
    
    print(f"\nProcessing file: {filename}")
    print(f"Configuration: bucket_size={bucket_size}, max_intervals={max_intervals}")
    print("="*50)
    
    # Process first file
    print(f"\nProcessing folder A: {folder_a}")
    start_times_a = read_txt_and_extract_start_times(file_a_path)
    if not start_times_a:
        print(f"No data found in {file_a_path}")
        return
    
    intervals_a = calculate_intervals(start_times_a)
    if not intervals_a:
        print(f"Cannot calculate intervals for {file_a_path}")
        return
    
    stats_a = analyze_intervals_by_buckets(intervals_a, bucket_size)
    
    # Process second file
    print(f"\nProcessing folder B: {folder_b}")
    start_times_b = read_txt_and_extract_start_times(file_b_path)
    if not start_times_b:
        print(f"No data found in {file_b_path}")
        return
    
    intervals_b = calculate_intervals(start_times_b)
    if not intervals_b:
        print(f"Cannot calculate intervals for {file_b_path}")
        return
    
    stats_b = analyze_intervals_by_buckets(intervals_b, bucket_size)
    
    # Create comparison plot
    fig, ax = plt.subplots(1, 1, figsize=(12, 8))
    fig.suptitle(f'Cumulative Distribution Comparison: {filename}', fontsize=18, fontweight='bold')
    
    # 创建严格限制在 max_intervals 范围内的横坐标
    x_positions = list(range(max_intervals))  # 0 到 max_intervals-1
    
    # Plot first dataset (严格限制到 max_intervals)
    if stats_a and 'cumulative_percentages' in stats_a:
        cum_percentages_a = list(stats_a['cumulative_percentages'].values())
        
        # 创建恰好 max_intervals 个数据点
        filtered_percentages_a = []
        for i in range(max_intervals):
            if i < len(cum_percentages_a):
                filtered_percentages_a.append(cum_percentages_a[i])
            else:
                # 如果数据不够，使用最后一个可用的百分比值
                filtered_percentages_a.append(cum_percentages_a[-1] if cum_percentages_a else 0)
        
        ax.plot(x_positions, filtered_percentages_a, 'o-', linewidth=3, markersize=6, 
                color='#2ecc71', label=f'Folder A ({os.path.basename(folder_a)})')
    
    # Plot second dataset (严格限制到 max_intervals)
    if stats_b and 'cumulative_percentages' in stats_b:
        cum_percentages_b = list(stats_b['cumulative_percentages'].values())
        
        # 创建恰好 max_intervals 个数据点
        filtered_percentages_b = []
        for i in range(max_intervals):
            if i < len(cum_percentages_b):
                filtered_percentages_b.append(cum_percentages_b[i])
            else:
                # 如果数据不够，使用最后一个可用的百分比值
                filtered_percentages_b.append(cum_percentages_b[-1] if cum_percentages_b else 0)
        
        ax.plot(x_positions, filtered_percentages_b, 's-', linewidth=3, markersize=6, 
                color='#e74c3c', label=f'Folder B ({os.path.basename(folder_b)})')
    
    # 设置图表属性，严格限制横坐标范围
    ax.set_title(f'Cumulative Distribution Percentage (Displaying {max_intervals} intervals)', fontsize=14, fontweight='bold')
    ax.set_xlabel(f'Interval Index (Each interval = {bucket_size} units, Range: 0-{max_intervals-1})')
    ax.set_ylabel('Cumulative Percentage (%)')
    
    # 严格设置 x 轴范围：从 0 到 max_intervals-1
    ax.set_xlim(0, max_intervals - 1)
    ax.set_ylim(0, 100)
    
    # 设置 x 轴刻度，确保不超过 max_intervals
    if max_intervals <= 50:
        # 如果间隔数较少，显示更多刻度
        tick_interval = max(1, max_intervals // 10)
        x_ticks = list(range(0, max_intervals, tick_interval))
        if (max_intervals - 1) not in x_ticks:
            x_ticks.append(max_intervals - 1)
    else:
        # 如果间隔数较多，显示较少刻度
        tick_interval = max(1, max_intervals // 10)
        x_ticks = list(range(0, max_intervals, tick_interval))
        if (max_intervals - 1) not in x_ticks:
            x_ticks.append(max_intervals - 1)
    
    ax.set_xticks(x_ticks)
    ax.grid(True, alpha=0.3)
    ax.legend()
    
    # 添加一些统计信息到图表上
    textstr = f'Max display range: 0-{(max_intervals-1)*bucket_size} units\nBucket size: {bucket_size} units'
    props = dict(boxstyle='round', facecolor='wheat', alpha=0.8)
    ax.text(0.02, 0.98, textstr, transform=ax.transAxes, fontsize=10,
            verticalalignment='top', bbox=props)
    
    # Save the plot
    plt.tight_layout()
    plt.savefig(output_image_name, dpi=300, bbox_inches='tight')
    plt.show()
    
    print(f"\nComparison plot saved as: {output_image_name}")
    print(f"X-axis strictly limited to: 0 to {max_intervals-1} (total {max_intervals} points)")
    
    # Print summary comparison
    print_comparison_summary(intervals_a, intervals_b, stats_a, stats_b, bucket_size, max_intervals, folder_a, folder_b)
def print_comparison_summary(intervals_a, intervals_b, stats_a, stats_b, bucket_size, max_intervals, folder_a, folder_b):
    """
    Print comparison summary statistics
    
    Args:
    intervals_a, intervals_b: lists of time intervals
    stats_a, stats_b: statistical analysis results
    bucket_size: size of each bucket
    max_intervals: maximum number of intervals displayed
    folder_a, folder_b: folder names for labeling
    """
    print(f"\n{'='*80}")
    print("COMPARISON SUMMARY")
    print(f"Bucket size: {bucket_size} units")
    print(f"Display range: 0 to {max_intervals * bucket_size - 1} units ({max_intervals} intervals)")
    print(f"{'='*80}")
    
    # Basic statistics comparison
    print(f"\nBasic Statistics Comparison:")
    print(f"{'Metric':<20} {'Folder A':<15} {'Folder B':<15} {'Difference'}")
    print("-" * 65)
    
    if intervals_a and intervals_b:
        metrics = [
            ("Total intervals", len(intervals_a), len(intervals_b)),
            ("Mean interval", np.mean(intervals_a), np.mean(intervals_b)),
            ("Median interval", np.median(intervals_a), np.median(intervals_b)),
            ("Std deviation", np.std(intervals_a), np.std(intervals_b)),
            ("Min interval", min(intervals_a), min(intervals_b)),
            ("Max interval", max(intervals_a), max(intervals_b))
        ]
        
        for metric_name, val_a, val_b in metrics:
            diff = val_b - val_a
            print(f"{metric_name:<20} {val_a:<15.2f} {val_b:<15.2f} {diff:+.2f}")
    
    # Key threshold statistics
    if intervals_a and intervals_b:
        # Generate threshold list based on bucket_size
        thresholds = []
        for i in range(1, 11):  # Generate first 10 thresholds
            threshold = i * bucket_size * 10  # Every 10 buckets
            if threshold <= max_intervals * bucket_size:
                thresholds.append(threshold)
        
        # Add some common thresholds if they make sense
        common_thresholds = [100, 200, 300, 500, 1000]
        for threshold in common_thresholds:
            if threshold <= max_intervals * bucket_size and threshold not in thresholds:
                thresholds.append(threshold)
        
        thresholds = sorted(set(thresholds))
        
        if thresholds:
            print(f"\nKey Threshold Comparison:")
            print(f"{'Threshold':<12} {'Folder A':<20} {'Folder B':<20} {'Difference'}")
            print("-" * 75)
            
            for threshold in thresholds:
                count_a = sum(1 for interval in intervals_a if interval < threshold)
                count_b = sum(1 for interval in intervals_b if interval < threshold)
                percentage_a = (count_a / len(intervals_a)) * 100
                percentage_b = (count_b / len(intervals_b)) * 100
                diff_percentage = percentage_b - percentage_a
                
                print(f"< {threshold:<10} {count_a:>6} ({percentage_a:>5.1f}%) {count_b:>6} ({percentage_b:>5.1f}%) {diff_percentage:+6.1f}%")

def print_summary_statistics(intervals, stats, label=""):
    """
    Print summary statistics
    
    Args:
    intervals: list of time intervals
    stats: statistical analysis results
    label: label for the dataset
    """
    print(f"\n{label} Statistics Summary:")
    print("="*40)
    
    if intervals:
        print(f"Basic Statistics:")
        print(f"   Total interval count: {len(intervals):,}")
        print(f"   Mean interval: {np.mean(intervals):,.2f}")
        print(f"   Median interval: {np.median(intervals):,.2f}")
        print(f"   Standard deviation: {np.std(intervals):,.2f}")
        print(f"   Min interval: {min(intervals):,}")
        print(f"   Max interval: {max(intervals):,}")
        print(f"   Interval range: {max(intervals) - min(intervals):,}")

def main():
    """
    Main function: execute complete comparison analysis workflow
    """
    # 用户可以在这里自定义参数
    folder_a = "upim"
    folder_b = "mmu-2"
    bucket_size = 5         # 每个桶的大小（单位）
    max_intervals = 200     # 显示的最大间隔数量
    
    print("DMA Start Time Interval Comparison Analysis")
    print("="*60)
    print(f"Configuration:")
    print(f"  - Folder A: {folder_a}")
    print(f"  - Folder B: {folder_b}")
    print(f"  - Bucket size: {bucket_size} units")
    print(f"  - Maximum intervals to display: {max_intervals}")
    print(f"  - Display range: 0 to {max_intervals * bucket_size - 1} units")
    print("="*60)
    
    # Create interval folder if it doesn't exist
    interval_folder = "interval"
    if not os.path.exists(interval_folder):
        os.makedirs(interval_folder)
        print(f"Created folder: {interval_folder}")
    
    # Find matching files
    matching_files = find_matching_files(folder_a, folder_b)
    
    if not matching_files:
        print("No matching txt files found in both folders")
        return
    
    # Process each matching file
    for filename in matching_files:
        print(f"\n{'='*60}")
        print(f"Processing: {filename}")
        print(f"{'='*60}")
        
        # Create output filename: remove .txt and add .png, save in interval folder
        base_filename = os.path.splitext(filename)[0]  # Remove .txt extension
        output_image_path = os.path.join(interval_folder, f"{base_filename}_{max_intervals}intervals.png")
        
        create_comparison_visualization(folder_a, folder_b, filename, output_image_path, bucket_size, max_intervals)

# 额外添加一个便捷函数，可以直接调用来指定参数
def compare_with_custom_settings(folder_a="upim", folder_b="mmu-2", bucket_size=10, max_intervals=100):
    """
    便捷函数：使用自定义设置进行对比分析
    
    Args:
    folder_a: 第一个文件夹路径，默认"upim"
    folder_b: 第二个文件夹路径，默认"mmu-2"
    bucket_size: 桶大小，默认10
    max_intervals: 最大显示间隔数，默认100
    """
    print("DMA Start Time Interval Comparison with Custom Settings")
    print("="*60)
    print(f"Configuration:")
    print(f"  - Folder A: {folder_a}")
    print(f"  - Folder B: {folder_b}")
    print(f"  - Bucket size: {bucket_size} units")
    print(f"  - Maximum intervals to display: {max_intervals}")
    print(f"  - Display range: 0 to {max_intervals * bucket_size - 1} units")
    print("="*60)
    
    # Create interval folder if it doesn't exist
    interval_folder = "interval"
    if not os.path.exists(interval_folder):
        os.makedirs(interval_folder)
        print(f"Created folder: {interval_folder}")
    
    # Find matching files
    matching_files = find_matching_files(folder_a, folder_b)
    
    if not matching_files:
        print("No matching txt files found in both folders")
        return
    
    # Process each matching file
    for filename in matching_files:
        print(f"\n{'='*60}")
        print(f"Processing: {filename}")
        print(f"{'='*60}")
        
        # Create output filename
        base_filename = os.path.splitext(filename)[0]
        output_image_path = os.path.join(interval_folder, f"{base_filename}_{max_intervals}intervals.png")
        
        create_comparison_visualization(folder_a, folder_b, filename, output_image_path, bucket_size, max_intervals)
    
    print("\nAnalysis complete!")

if __name__ == "__main__":
#main()
    
    # 示例：如果你想使用不同的设置，可以取消注释下面的行
    compare_with_custom_settings(folder_a="upim", folder_b="mmu-2", bucket_size=10, max_intervals=80)
    # compare_with_custom_settings(folder_a="upim", folder_b="mmu-2", bucket_size=20, max_intervals=50)
```
